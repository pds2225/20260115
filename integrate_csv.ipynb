```

import pandas as pd

DATA = "/workspace/value_up_ai/data"

# 기존 buyer_db.csv 에서 진짜 데이터만 남기기 (Customs_* 소스만 = 55행)
df_old = pd.read_csv(f"{DATA}/buyer_db.csv", dtype=str)
df_real = df_old[df_old['source'].str.startswith('Customs_', na=False)].copy()
print(f"기존 진짜 행(Customs_*): {len(df_real)}행")

# KOTRA SNS 바이어 → buyer_db 스키마로 변환
df_sns = pd.read_csv(f"{DATA}/kotra_sns_buyers.csv", dtype=str)

# HS코드 있는 것만 (buyer_db는 HS코드 기반)
df_sns_hs = df_sns[df_sns['hs_code'].notna() & df_sns['hs_code'].str.strip().str.len().gt(3)].copy()
df_sns_hs['hs_code'] = df_sns_hs['hs_code'].str.strip().str[:6]  # 6자리로

# buyer_db 스키마에 맞춰 컬럼 생성
df_sns_mapped = pd.DataFrame({
    'hs_code':    df_sns_hs['hs_code'],
    'country':    df_sns_hs['country'],
    'buyer_name': df_sns_hs['buyer_name'],
    'annual_usd': '',
    'shipments':  '',
    'last_date':  '',
    'buyer_type': 'Importer',
    'city':       df_sns_hs['city'],
    'source':     'KOTRA_SNS_2025',
})

# 기존 진짜 행 + KOTRA SNS 합치기
df_merged = pd.concat([df_real, df_sns_mapped], ignore_index=True)
df_merged.to_csv(f"{DATA}/buyer_db.csv", index=False, encoding='utf-8-sig')

print(f"✅ buyer_db.csv 교체 완료")
print(f"  - 기존 진짜(Customs_*): {len(df_real)}행")
print(f"  - KOTRA SNS 실데이터: {len(df_sns_mapped)}행")
print(f"  - 최종 합계: {len(df_merged)}행")
print(f"\n국가 분포 Top10:")
print(df_merged['country'].value_counts().head(10))
```
