```

import requests

API_KEY = "83b96790de580e57527e049d59bfcb18ae34d2bfe646c11a5d2ee6b3d95e9b23"
BASE = "https://apis.data.go.kr/B552696/buyer/getBuyerList"

# 110=아프가니스탄, 115=방글라데시, 120=대만, 130=홍콩, 140=일본
# 아시아 주요국: 베트남, 태국, 싱가포르, 말레이시아, 인도네시아, 인도, 중국, 호주
# 135 = ? (캄보디아?), 170 = ?, 175 = ?...

ASIA_CANDIDATES = list(range(100, 210, 1))  # 아시아 구역 전수 조사

print("아시아 구역 전수 탐색...")
found = {}
for code in ASIA_CANDIDATES:
    try:
        r = requests.get(BASE, params={
            "serviceKey": API_KEY, "pageNo": "1", "numOfRows": "1",
            "ctryCd": str(code), "prodNm": "trade", "_type": "json"
        }, timeout=4)
        d = r.json()
        body = d.get("response", {}).get("body")
        if body:
            items = body.get("items", {}).get("item", [])
            if items:
                ctry_nm = items[0].get("ctryNm", "")
                found[str(code)] = ctry_nm
    except:
        pass

print("아시아 구역 발견:")
for code, name in sorted(found.items(), key=lambda x: int(x[0])):
    print(f"  ctryCd={code} → {name}")
```
